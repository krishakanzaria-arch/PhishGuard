```python

import pandas as pd

df = pd.read_csv("dataset/CEAS_08.csv/CEAS_08.csv")

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())

print("\nMissing values:")
print(df.isnull().sum())
```

```python

# 1. Check how many emails belong to each class
print("Email label counts:")
print(df["label"].value_counts(dropna=False))

# 2. Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

# 3. Check for duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# 4. Check the data types
print("\nColumn data types:")
print(df.dtypes)
```

```python

# Make a working copy so the original data stays unchanged
data = df.copy()

# Fill missing subjects with empty text
data["subject"] = data["subject"].fillna("")

# Make sure email fields are strings
data["subject"] = data["subject"].astype(str)
data["body"] = data["body"].astype(str)

# Combine subject and body into one text field
data["email_text"] = (
    data["subject"].str.strip() + " " +
    data["body"].str.strip()
).str.strip()

# Remove rows that contain no email text
data = data[data["email_text"].str.len() > 0].copy()

print("Original rows:", len(df))
print("Rows after cleaning:", len(data))
print("\nLabel distribution:")
print(data["label"].value_counts())
print("\nExample email text:")
print(data["email_text"].iloc[0][:500])
```

```python

from sklearn.model_selection import train_test_split

# Input: email text
X = data["email_text"]

# Target: email label
y = data["label"]

# Split into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training emails:", len(X_train))
print("Testing emails:", len(X_test))

print("\nTraining label distribution:")
print(y_train.value_counts())

print("\nTesting label distribution:")
print(y_test.value_counts())
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer

# Create our text-to-number converter
tfidf = TfidfVectorizer(
    lowercase=True,
    max_features=20000,
    ngram_range=(1, 2),
    stop_words="english"
)

# Learn vocabulary from training emails and convert them to numbers
X_train_tfidf = tfidf.fit_transform(X_train)

# Convert test emails using the same learned vocabulary
X_test_tfidf = tfidf.transform(X_test)

print("Training feature matrix:", X_train_tfidf.shape)
print("Testing feature matrix:", X_test_tfidf.shape)
print("Number of learned words/phrases:", len(tfidf.vocabulary_))
```

```python

from sklearn.linear_model import LogisticRegression

# Create the model
model = LogisticRegression(max_iter=1000, random_state=42)

# Train it using the training emails and their labels
model.fit(X_train_tfidf, y_train)

print("Model training completed successfully!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Predict labels for emails the model hasn't seen
y_pred = model.predict(X_test_tfidf)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print(f"Test accuracy: {accuracy:.4f}")
print(f"Test accuracy percentage: {accuracy * 100:.2f}%")

print("\nClassification report:")
print(classification_report(y_test, y_pred))

print("\nConfusion matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Create and train the Naive Bayes model
nb_model = MultinomialNB()
nb_model.fit(X_train_tfidf, y_train)

# Predict labels for the test emails
nb_pred = nb_model.predict(X_test_tfidf)

# Evaluate the model
nb_accuracy = accuracy_score(y_test, nb_pred)

print("Naive Bayes Results")
print("-------------------")
print(f"Test accuracy: {nb_accuracy:.4f}")
print(f"Test accuracy percentage: {nb_accuracy * 100:.2f}%")
print("\nClassification report:")
print(classification_report(y_test, nb_pred))
```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report

# Create and train the SVM model
svm_model = LinearSVC(random_state=42)
svm_model.fit(X_train_tfidf, y_train)

# Predict labels for the test emails
svm_pred = svm_model.predict(X_test_tfidf)

# Evaluate the model
svm_accuracy = accuracy_score(y_test, svm_pred)

print("Linear SVM Results")
print("------------------")
print(f"Test accuracy: {svm_accuracy:.4f}")
print(f"Test accuracy percentage: {svm_accuracy * 100:.2f}%")
print("\nClassification report:")
print(classification_report(y_test, svm_pred))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Compare the accuracy of all three models
comparison = pd.DataFrame({
    "Algorithm": [
        "Logistic Regression",
        "Multinomial Naive Bayes",
        "Linear SVM"
    ],
    "Accuracy (%)": [
        accuracy * 100,
        nb_accuracy * 100,
        svm_accuracy * 100
    ]
})

comparison = comparison.sort_values(
    by="Accuracy (%)",
    ascending=False
).reset_index(drop=True)

print("MODEL COMPARISON")
display(comparison.round(2))

# Plot the comparison
plt.figure(figsize=(8, 5))
plt.bar(comparison["Algorithm"], comparison["Accuracy (%)"])
plt.ylabel("Test Accuracy (%)")
plt.xlabel("Machine Learning Algorithm")
plt.title("PhishGuard: Model Accuracy Comparison")
plt.ylim(95, 100)
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python

# ROC-AUC ANALYSIS

import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

from sklearn.metrics import (
    roc_curve,
    roc_auc_score,
    auc
)

# Prepare email text
roc_df = df.copy()

roc_df["subject"] = roc_df["subject"].fillna("").astype(str)
roc_df["body"] = roc_df["body"].fillna("").astype(str)

roc_df["email_text"] = (
    roc_df["subject"] + " " + roc_df["body"]
).str.strip()

X = roc_df["email_text"]
y = roc_df["label"].astype(int)

# Split dataset into training and testing
X_train_roc, X_test_roc, y_train_roc, y_test_roc = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# TF-IDF feature extraction
roc_vectorizer = TfidfVectorizer(
    max_features=20000,
    stop_words="english"
)

X_train_tfidf_roc = roc_vectorizer.fit_transform(X_train_roc)
X_test_tfidf_roc = roc_vectorizer.transform(X_test_roc)

# Initialize models
roc_models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Naive Bayes": MultinomialNB(),
    "Linear SVM": LinearSVC(max_iter=5000)
}

# Plot ROC curves
plt.figure(figsize=(10, 7))

roc_results = []

for name, model in roc_models.items():

    # Train model
    model.fit(X_train_tfidf_roc, y_train_roc)

    # Get continuous prediction scores
    if hasattr(model, "predict_proba"):
        y_scores = model.predict_proba(X_test_tfidf_roc)[:, 1]
    else:
        y_scores = model.decision_function(X_test_tfidf_roc)

    # Calculate ROC curve and AUC
    fpr, tpr, thresholds = roc_curve(y_test_roc, y_scores)

    roc_auc = roc_auc_score(y_test_roc, y_scores)

    roc_results.append({
        "Algorithm": name,
        "ROC-AUC Score": roc_auc
    })

    # Plot curve
    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=f"{name} (AUC = {roc_auc:.4f})"
    )

# Random classifier reference line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.title("ROC Curves - PhishGuard Classification Models")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend(loc="lower right")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Display ROC-AUC comparison
roc_results_df = pd.DataFrame(roc_results)

display(
    roc_results_df.sort_values(
        by="ROC-AUC Score",
        ascending=False
    ).reset_index(drop=True)
)
```
